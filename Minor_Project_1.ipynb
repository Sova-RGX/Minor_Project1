```python
import numpy as np
#connecting the ide with the file to read it :
with open("hostel_bois.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()                                #the code here is debugged with the help of ai
print("The number of lines in the file :",len(lines))
```

```python
print("The number of lines in the file : 3178")
#To check the last line :
print(lines[-5:])
```

```python
for line in lines[:10]:
    print(line.strip())
```

```python
# Removing any blank spaces if any :
messages = []

for line in lines:
    line = line.strip()

    if " - " not in line:
        continue

    date_time, content = line.split(" - ", 1)

    if ": " in content:
        sender, text = content.split(": ", 1)

        messages.append({
            "datetime": date_time,
            "sender": sender,
            "text": text
        })
print(messages[:5])
print("Parsed messages:", len(messages))
```

```python
#Code-Block to identify participants who also send message

participants = set()

for m in messages:
    participants.add(m["sender"])

print(participants)
print("Participants:", len(participants))       #participant list
```

```python
#To check how many messsages each sender has sent

message_counts = {}

for message in messages:
    sender = message["sender"]

    if sender not in message_counts:
        message_counts[sender] = 0

    message_counts[sender] += 1

print(message_counts)
```

```python
#To find out the Start Date & The End Date of messages:

dates = []

for message in messages:
    date = message["datetime"].split(", ")[0]
    dates.append(date)

print("Start date:", dates[0])
print("End date:", dates[-1])
print("Number of days:", len(set(dates)))
```

```python
#To find out the days of the most activity and messaging in the group :

day_counts = {}

for message in messages:
    date = message["datetime"].split(", ")[0]

    if date not in day_counts:
        day_counts[date] = 0

    day_counts[date] += 1

most_active_day = max(day_counts, key=day_counts.get)

print("Most active day:", most_active_day)
print("Messages:", day_counts[most_active_day])
```

```python
#To find out the most active hours of the group :

hour_counts = {}

for message in messages:
    time = message["datetime"].split(", ")[1]
    hour = int(time.split(":")[0])

    if hour not in hour_counts:
        hour_counts[hour] = 0

    hour_counts[hour] += 1

most_active_hour = max(hour_counts, key=hour_counts.get)

print("Most active hour:", most_active_hour)
print("Messages:", hour_counts[most_active_hour])
```

```python
#GROUP OVERVIEW... :

total_messages = len(messages)

sorted_counts = sorted(message_counts.items(), key=lambda item: item[1], reverse=True)

print("============================================================")
print(" GROUP OVERVIEW")
print("============================================================")
print(" Group           : Hostel Bois 4ever")
print(" Period          : 01 April 2024 to 30 May 2024 (60 days)")
print(" Total messages  :", total_messages)
print(" Participants    :", len(participants))

print("\n MESSAGES PER PERSON")

for sender, count in sorted_counts:
    percentage = (count / total_messages) * 100
    print(f" {sender:<7}: {count:>3} ({percentage:.1f}%)")
```

```python
# Feature 4 - heat map
# Create 6 x 24 heatmap
heatmap = np.zeros((len(participants), 24), dtype=int)

participant_list = sorted(participants)

# Fill the heatmap
for message in messages:
    sender = message["sender"]
    time = message["datetime"].split(", ")[1]
    hour = int(time.split(":")[0])

    row = participant_list.index(sender)

    heatmap[row][hour] += 1


# Display the heatmap
print("=" * 60)
print(" ACTIVITY HEATMAP")
print("=" * 60)

print("       " + " ".join("{:02d}".format(h) for h in range(0, 24, 3)))

for i in range(len(participant_list)):

    person = participant_list[i]
    row = heatmap[i]
    maximum = max(row)

    print("" + person.ljust(8), end="")

    for h in range(0, 24, 3):

        value = max(row[h:h + 3])

        if value == 0:
            symbol = "."
        elif value <= maximum * 0.25:
            symbol = "░"
        elif value <= maximum * 0.50:
            symbol = "▒"
        else:
            symbol = "█"

        print(symbol,end="  ")

    print()

print("=" * 60)
```

```python
#Feature 5 - Top 10 words & groups favourite words
word_counts = {}

stop_words = ["i", "is", "the", "a", "and", "or", "to", "of", "in", "on", "for"]

for message in messages:

    text = message["text"]

    #here we are skipping media and deleted messages
    if text == "<Media omitted>" or text == "This message was deleted":
        continue

    text = text.lower()
    words = text.split()

    for word in words:

        word = word.strip(".,!?;:()[]\"'")

        if word == "":
            continue

        if word in stop_words:
            continue

        if word not in word_counts:
            word_counts[word] = 0

        word_counts[word] += 1
sorted_words = sorted(word_counts.items(), key=lambda item: item[1], reverse=True)

top_words = sorted_words[:10]
print("=" * 60)
print("TOP 10 WORDS")
print("=" * 60)
for word, count in top_words:
    print(f"{word:<15} {count}")
maximum = top_words[0][1]
print("=" * 60)
print("\n")
print("=" * 60)
print("THIS GROUP'S FAVOURITE WORDS")
print("=" * 60)
for word, count in top_words:

    bar_length = int((count / maximum) * 20)
    bar = "█" * bar_length

    print(f"{word:<15} {bar:<20} {count}")
print("=" * 60)
```

```python
# Feature no.6
from datetime import datetime

response_gaps = {}

for person in participants:
    response_gaps[person] = []

previous_time = None
previous_sender = None

for message in messages:

    current_time = datetime.strptime(
        message["datetime"],
        "%d/%m/%y, %H:%M"
    )

    current_sender = message["sender"]

    if previous_time is not None and current_sender != previous_sender:

        gap = current_time - previous_time
        gap_seconds = gap.total_seconds()

        response_gaps[current_sender].append(gap_seconds)

    previous_time = current_time
    previous_sender = current_sender


average_response_time = {}

for person in response_gaps:

    gaps = response_gaps[person]

    if len(gaps) > 0:
        average = sum(gaps) / len(gaps)
    else:
        average = 0

    average_response_time[person] = average


average_response_minutes = {}

for person in average_response_time:

    seconds = average_response_time[person]
    minutes = seconds / 60

    average_response_minutes[person] = minutes


active_days = {}

for person in participants:
    active_days[person] = set()

for message in messages:

    person = message["sender"]
    date = message["datetime"].split(", ")[0]

    active_days[person].add(date)


silent_streaks = {}

for person in participants:

    dates = []

    for date in active_days[person]:

        real_date = datetime.strptime(date, "%d/%m/%y")
        dates.append(real_date)

    dates.sort()

    longest = 0

    for i in range(len(dates) - 1):

        gap = (dates[i + 1] - dates[i]).days

        if gap > 1:

            silent_days = gap - 1

            if silent_days > longest:
                longest = silent_days

    silent_streaks[person] = longest


# Feature 6 final output

fastest_person = min(
    average_response_minutes,
    key=average_response_minutes.get
)

slowest_person = max(
    average_response_minutes,
    key=average_response_minutes.get
)


print("=" * 60)
print(" RESPONSE PATTERNS")
print("=" * 60)

print(
    f" Fastest replier : {fastest_person} "
    f"(avg {average_response_minutes[fastest_person]:.1f} minutes)"
)

slowest_minutes = average_response_minutes[slowest_person]

if slowest_minutes >= 60:

    slowest_hours = slowest_minutes / 60

    print(
        f" Slowest replier : {slowest_person} "
        f"(avg {slowest_hours:.1f} hours)"
    )

else:

    print(
        f" Slowest replier : {slowest_person} "
        f"(avg {slowest_minutes:.1f} minutes)"
    )


print("\n LONGEST SILENT STREAKS (consecutive days with zero messages)")

sorted_streaks = sorted(
    silent_streaks.items(),
    key=lambda item: item[1],
    reverse=True
)

for person, days in sorted_streaks:

    if days == 1:
        print(f" {person:<7}: {days} day")
    else:
        print(f" {person:<7}: {days} days")

print("=" * 60)
```

```python
#Feature 7 - Personality Archetypes of the group members
from datetime import datetime


def is_real_text(message):
    text = message["text"].strip()

    if text == "<Media omitted>":
        return False

    if text == "This message was deleted":
        return False

    return True



def detect_spammer(person):
    bursts = []
    current_burst = 0

    for message in messages:
        if message["sender"] == person:
            current_burst += 1
        else:
            if current_burst > 0:
                bursts.append(current_burst)

            current_burst = 0

    if current_burst > 0:
        bursts.append(current_burst)

    if len(bursts) == 0:
        return 0

    return sum(bursts) / len(bursts)



def detect_group_mom(person):
    caring_words = [
        "okay",
        "safe",
        "eat",
        "sleep",
        "take care",
        "are you",
        "please",
        "reminder",
        "drink water",
        "don't forget"
    ]

    total_messages = 0
    caring_messages = 0

    for message in messages:
        if message["sender"] == person:
            total_messages += 1
            text = message["text"].lower()

            for word in caring_words:
                if word in text:
                    caring_messages += 1
                    break

    if total_messages == 0:
        return 0, 0

    percentage = (caring_messages / total_messages) * 100

    return caring_messages, percentage


def detect_night_owl(person):
    total_messages = 0
    night_messages = 0

    for message in messages:
        if message["sender"] == person:
            total_messages += 1

            time = message["datetime"].split(", ")[1]
            hour = int(time.split(":")[0])

            if hour >= 23 or hour <= 4:
                night_messages += 1

    if total_messages == 0:
        return 0, 0

    percentage = (night_messages / total_messages) * 100

    return night_messages, percentage

def detect_storyteller(person):
    total_words = 0
    total_messages = 0

    for message in messages:
        if message["sender"] == person and is_real_text(message):
            total_words += len(message["text"].split())
            total_messages += 1

    if total_messages == 0:
        return 0

    return total_words / total_messages


def detect_drama_queen(person):
    total_messages = 0
    all_caps_messages = 0
    exclamation_messages = 0

    for message in messages:
        if message["sender"] == person and is_real_text(message):

            total_messages += 1
            text = message["text"]

            letters = []

            for character in text:
                if character.isalpha():
                    letters.append(character)

            if len(letters) >= 3:

                all_caps = True

                for character in letters:
                    if not character.isupper():
                        all_caps = False
                        break

                if all_caps:
                    all_caps_messages += 1

            if text.count("!") >= 2:
                exclamation_messages += 1

    if total_messages == 0:
        return 0, 0

    caps_percentage = (all_caps_messages / total_messages) * 100
    exclamation_percentage = (exclamation_messages / total_messages) * 100

    return caps_percentage, exclamation_percentage



def detect_ghost(person):
    active_dates = set()

    for message in messages:
        if message["sender"] == person:

            date_text = message["datetime"].split(", ")[0]

            date = datetime.strptime(
                date_text,
                "%d/%m/%y"
            ).date()

            active_dates.add(date)

    silent_days = 60 - len(active_dates)
    silent_percentage = (silent_days / 60) * 100

    return silent_days, silent_percentage



def detect_comedian(person):
    humor_words = [
        "lol",
        "lmao",
        "haha",
        "rofl",
        "lmfao"
    ]

    total_messages = 0
    funny_messages = 0

    for message in messages:
        if message["sender"] == person and is_real_text(message):

            total_messages += 1
            text = message["text"].lower()

            for word in humor_words:
                if word in text:
                    funny_messages += 1
                    break

    if total_messages == 0:
        return 0, 0

    percentage = (funny_messages / total_messages) * 100

    return funny_messages, percentage


def detect_question_master(person):
    total_messages = 0
    question_messages = 0

    for message in messages:
        if message["sender"] == person and is_real_text(message):

            total_messages += 1

            if message["text"].rstrip().endswith("?"):
                question_messages += 1

    if total_messages == 0:
        return 0, 0

    percentage = (question_messages / total_messages) * 100

    return question_messages, percentage


spammer_scores = {}
group_scores = {}
night_scores = {}
storyteller_scores = {}
drama_scores = {}
ghost_scores = {}
comedian_scores = {}
question_scores = {}


for person in participant_list:

    spammer_scores[person] = detect_spammer(person)

    group_scores[person] = detect_group_mom(person)

    night_scores[person] = detect_night_owl(person)

    storyteller_scores[person] = detect_storyteller(person)

    drama_scores[person] = detect_drama_queen(person)

    ghost_scores[person] = detect_ghost(person)

    comedian_scores[person] = detect_comedian(person)

    question_scores[person] = detect_question_master(person)



highest_group = 0

for person in participant_list:
    if group_scores[person][1] > highest_group:
        highest_group = group_scores[person][1]


highest_comedian = 0

for person in participant_list:
    if comedian_scores[person][1] > highest_comedian:
        highest_comedian = comedian_scores[person][1]



final_results = {}


for person in participant_list:

    candidates = []


    # SPAMMER
    if spammer_scores[person] > 3:
        candidates.append(
            ("THE SPAMMER", spammer_scores[person])
        )


    # GROUP MOM
    if group_scores[person][1] == highest_group:
        candidates.append(
            ("THE GROUP MOM", group_scores[person][1])
        )


    # NIGHT OWL
    if night_scores[person][1] > 60:
        candidates.append(
            ("THE NIGHT OWL", night_scores[person][1])
        )


    # STORYTELLER
    if storyteller_scores[person] > 30:
        candidates.append(
            ("THE STORYTELLER", storyteller_scores[person])
        )


    # DRAMA QUEEN
    if (
        drama_scores[person][0] > 30
        or drama_scores[person][1] > 0
    ):
        candidates.append(
            ("THE DRAMA QUEEN", drama_scores[person][0])
        )


    # GHOST
    if ghost_scores[person][1] > 60:
        candidates.append(
            ("THE GHOST", ghost_scores[person][1])
        )


    # COMEDIAN
    if comedian_scores[person][1] == highest_comedian:
        candidates.append(
            ("THE COMEDIAN", comedian_scores[person][1])
        )


    # QUESTION MASTER
    if question_scores[person][1] > 25:
        candidates.append(
            ("THE QUESTION MASTER", question_scores[person][1])
        )


    # SELECT ONE ARCHETYPE
    if len(candidates) == 0:

        final_results[person] = "NO CLEAR ARCHETYPE"

    else:

        best = candidates[0]

        for candidate in candidates[1:]:

            if candidate[1] > best[1]:
                best = candidate

        final_results[person] = best[0]


print("="*60)
print("FINAL PERSONALITY ARCHETYPES")
print("="*60)

for person in participant_list:
    print(person, "->", final_results[person])
print("="*60)
```
